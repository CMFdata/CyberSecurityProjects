*B. Describir la cadena de permisos*

La cadena de 10 caracteres puede descomponerse para determinar quién está autorizado a acceder al archivo y sus **permisos específicos**. Los personajes y lo que representan son los siguientes:
- Caracter 1º: Este carácter es o bien una d o un guion (-) e indica el tipo de archivo. Si es una d, es un **directorio**. Si es un guion (-), es un **archivo normal**.
- Caracteres 2º a 4º: Estos caracteres indican los permisos de **lectura** (r), **escritura** (w) y **ejecución** (x) para el usuario. Cuando uno de estos caracteres es un **guion** (-) en su lugar, indica que este permiso **no se concede al usuario**.
- Caracteres 5º a 7º: Estos caracteres indican los permisos de **lectura** (r), **escritura** (w) y **ejecución** (x) para el grupo. Cuando uno de estos caracteres es un **guion** (-) en su lugar, indica que este permiso **no se concede al grupo**.
- Caracteres 8º-10º: Estos caracteres indican los permisos de **lectura** (r), **escritura** (w) y **ejecución** (x) para otros. Este tipo de propietario consiste en todos los demás usuarios del sistema excepto el usuario y el grupo. Cuando uno de estos caracteres es un **guion** (-) en su lugar, eso indica que este permiso **no se concede para otros**.

Por ejemplo, los **permisos de archivo** para project_t.txt son **-rw-r--**. Dado que el primer carácter es un guion (-), esto indica que project_t.txt es un archivo, no un directorio. El segundo, quinto y octavo carácter son todos r, lo que indica que usuario, grupo y otros tienen **permisos de lectura**. El tercer y sexto carácter son w, lo que indica que solo el usuario y el grupo tienen **permisos de escritura**. Nadie posee **permisos de ejecución** para project_t.txt.
