*B. Reporte final de acceso no autorizado a datos de clientes*

###*Informe Final del Incidente*

**Resumen ejecutivo**

La organización experimentó un incidente de seguridad el 19 de Abril de 2026, a las 19:20 hora, durante el cual una persona pudo obtener acceso no autorizado a información personal identificable (PII) de clientes e información financiera. Se vieron afectados aproximadamente 50.000 registros de clientes. El impacto financiero del incidente se estima en 100.000 dólares en costes directos y posibles pérdidas de ingresos. El incidente ya está cerrado y se ha llevado a cabo una investigación exhaustiva.


**Cronología**

Aproximadamente a las 15:13 hora, el 20 de Abril de 2026, un empleado recibió un correo electrónico de una dirección externa. El remitente afirmó que había conseguido robar datos de clientes. A cambio de no facilitar los datos en foros públicos, el remitente solicitó un pago en criptomonedas de 25.000 dólares. El empleado asumió que el correo era spam y lo eliminó.
El 22 de diciembre de 2026, el mismo empleado recibió otro correo electrónico del mismo remitente. Este correo incluía una muestra de los datos robados de clientes y una demanda de pago aumentada de 50.000 dólares.
Ese mismo día, el empleado notificó al equipo de seguridad, que inició la investigación sobre el incidente. Entre el 22 y el 25 de Abril de 2026, el equipo de seguridad se centró en determinar cómo se robaron los datos y la magnitud del robo.


**Investigación**

El equipo de seguridad recibió la alerta y viajó al lugar para comenzar la investigación.
La causa raíz del incidente se identificó como una vulnerabilidad en la aplicación web de comercio electrónico. Esta vulnerabilidad permitía al atacante realizar un ataque de navegación forzada y acceder a los datos de transacciones del cliente modificando el número de pedido incluido en la cadena URL de una página de confirmación de compra. Esta vulnerabilidad permitía al atacante acceder a las páginas de confirmación de compra del cliente, exponiendo los datos del cliente, que luego recogía y exfiltraba.
Tras confirmar la vulnerabilidad de la aplicación web, el equipo de seguridad analizó los registros de acceso a la aplicación web. Los registros indicaban que el atacante accedió a la información de miles de páginas de confirmación de compras.

**Respuesta y remediación**

La organización colaboró con el departamento de relaciones públicas para divulgar la brecha de datos a sus clientes. Además, la organización ofreció servicios gratuitos de protección de identidad a los clientes afectados por el incidente.
Después de que el equipo de seguridad revisara los registros asociados del servidor web, la causa del ataque quedó muy clara. Había una única fuente de registro que mostraba un volumen excepcionalmente alto de pedidos de clientes listados secuencialmente.

**Recomendaciones**

Para evitar que se repitan en el futuro, estamos tomando las siguientes medidas:

●	Realizar escaneos rutinarios de vulnerabilidades y pruebas de penetración.

●	Implementa los siguientes mecanismos de control de acceso:

    ○	Implementa la lista de permisos para permitir el acceso a un conjunto específico de URLs y bloquea automáticamente todas las solicitudes fuera de este rango de URL.

    ○	Asegúrate de que solo los usuarios autenticados tengan acceso autorizado al contenido.
