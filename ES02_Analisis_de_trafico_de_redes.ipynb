*B. Informe de Incidente de Seguridad*

**Sección 1: Identificar el protocolo de red implicado en el incidente**

El protocolo implicado en el incidente es el protocolo de transferencia de hipertexto (HTTP). Como el problema era acceder al servidor web durante yummyrecipesforme.com, sabemos que las solicitudes a servidores web para páginas web implican tráfico http. Además, cuando ejecutamos tcpdump y accedimos a la web yummyrecipesforme.com el archivo de registro correspondiente de tcpdump mostró el uso del protocolo http al contactar con el servidor. Se observa que el archivo malicioso es transportado a los ordenadores de los usuarios usando el protocolo HTTP en la capa de aplicación.

**Sección 2: Documentar el incidente**

Varios clientes contactaron con el servicio de ayuda del sitio web indicando que, al visitarlo, se les pidió que descargaran y ejecutaran un archivo que contenía acceso a nuevas recetas. Sus ordenadores personales han funcionado lentamente desde entonces. El propietario de la web intentó iniciar sesión en el servidor web, pero notó que estaba bloqueado en su cuenta.

El analista de ciberseguridad utilizó un entorno sandbox para abrir la web sin afectar la red de la empresa. Luego, el analista ejecutaba tcpdump para capturar los paquetes de tráfico de red generados al interactuar con el sitio web. Se le pidió al analista que descargara un archivo que afirmaba que daría acceso a recetas gratuitas, aceptó la descarga y la ejecutó. El navegador redirigía entonces al analista a una web falsa (greatrecipesforme.com).

El analista de ciberseguridad inspeccionó el registro de tcpdump y observó que el navegador solicitó inicialmente la dirección IP del sitio web yummyrecipesforme.com. Una vez establecida la conexión con el sitio web a través del protocolo HTTP, el analista recordó haber descargado y ejecutado el archivo. Los registros mostraron un cambio repentino en el tráfico de red cuando el navegador solicitó una nueva dirección IP para la URL greatrecipesforme.com. El tráfico de red se redirigía entonces a la nueva dirección IP del sitio web greatrecipesforme.com.

El profesional senior de ciberseguridad analizó el código fuente de los sitios web y el archivo descargado. El analista descubrió que un atacante había manipulado la web para añadir código que incitaba a los usuarios a descargar un archivo malicioso disfrazado de actualización del navegador. Como el propietario del sitio web declaró que había sido bloqueado fuera de su cuenta de administrador, el equipo cree que el atacante utilizó un ataque de fuerza bruta para acceder a la cuenta y cambiar la contraseña de administrador. La ejecución del archivo malicioso comprometió los ordenadores de los usuarios finales.

**Sección 3: Recomendar una remediación para ataques de fuerza bruta**

Una medida de seguridad que el equipo planea implementar para protegerse de ataques de fuerza bruta es impedir que se usen contraseñas anteriores. Dado que la vulnerabilidad que llevó a este ataque fue la capacidad del atacante para usar una contraseña predeterminada para iniciar sesión, es importante que evitemos que se utilicen contraseñas antiguas, como las predeterminadas, para restablecer la contraseña. Otra medida de apoyo es exigir actualizaciones de contraseña más frecuentes, para que en caso de que alguna persona no autorizada se entere de la contraseña, sea menos probable que pueda usarla si se actualiza pronto. Finalmente, otra solución útil es implementar la autenticación en dos factores (2FA). La autenticación en 2FA requiere autenticación mediante contraseña y también confirmando un código de acceso único (OTP) enviado a su correo electrónico o teléfono. Una vez que el usuario confirma su identidad a través de sus credenciales de acceso y el OTP, tendrá acceso al sistema. Cualquier actor malicioso que intente un ataque por fuerza bruta probablemente no tendrá acceso al sistema porque requiere autenticación adicional.

*C. Evaluación de riesgos de seguridad*

**Herramientas y métodos de endurecimiento (hardering) a implementar**

Tres herramientas de endurecimiento que la organización puede utilizar para abordar las vulnerabilidades encontradas incluyen:
1.   Implementación de la autenticación multifactor (MFA)
2.   Establecimiento y aplicación de políticas de contraseñas fuertes
3.   Realizar mantenimiento de firewalls regularmente

La MFA requiere que los usuarios utilicen más de una forma de identificar y verificar sus credenciales antes de acceder a una aplicación. Algunos métodos de MFA incluyen escaneos de huellas dactilares, tarjetas de identidad, números PIN y contraseñas.

Las políticas de contraseñas pueden refinarse para incluir normas sobre la longitud de la contraseña, una lista de caracteres aceptables y un aviso para desalentar el intercambio de contraseñas. También pueden incluir reglas sobre intentos de inicio de sesión fallidos, como que el usuario pierda acceso a la red tras cinco intentos fallidos.

El mantenimiento del firewalls implica revisar y actualizar regularmente las configuraciones de seguridad para anticiparse a posibles amenazas.


**Recomendaciones**

Hacer cumplir la autenticación multifactor (**MFA**) añade una capa adicional de seguridad más allá de la simple contraseña. Esto reducirá la probabilidad de que un actor malicioso pueda acceder a una red mediante un ataque de fuerza bruta o relacionado, ya que se requiere un esfuerzo adicional para autenticarse de más de una manera. La MFA también puede reducir la probabilidad de que las personas compartan contraseñas. Dado que el destinatario de la contraseña compartida necesitaría poseer autenticación adicional además de la contraseña, la MFA hace menos útil compartir contraseñas, lo que hace menos probable que se compartan.

Crear y hacer cumplir una política de contraseñas dentro de la empresa hará que sea cada vez más difícil para los actores maliciosos acceder a la red. Políticas como suspender la cuenta tras cierto número de inicios de sesión pueden evitar ataques de fuerza bruta exitosos. Aumentar la complejidad de las contraseñas, requerir actualizaciones más frecuentes y no permitir que se reutilicen las contraseñas también ayuda a frenar la infiltración de actores maliciosos en la red.

El mantenimiento del firewall debería hacerse de forma regular. Los administradores de red deben asegurarse de que existan normas de firewalls que reflejen los estándares más actualizados para el tráfico permitido y denegado. El tráfico procedente de fuentes sospechosas debe incluirse en una lista de tráfico denegado. Las reglas del firewalls deben actualizarse siempre que ocurra un evento de seguridad, especialmente un evento que permita la entrada de tráfico sospechoso en la red. Esta medida puede usarse para proteger contra diversos ataques DoS (Denial-of-Service) y DDoS (Distributed Denial-of-Service).
