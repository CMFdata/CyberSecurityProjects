*B. Security Incident Report*

**Section 1: Identify the network protocol involved in the incident**

The protocol involved in the incident is the Hypertext Transfer Protocol (HTTP). Since the problem was accessing the web server for yummyrecipesforme.com, we know that requests to web servers for web pages involve http traffic. Furthermore, when we ran tcpdump and accessed the yummyrecipesforme.com website, the corresponding tcpdump log file showed the use of the http protocol when contacting the server. It is observed that the malicious file is transported to users' computers using the HTTP protocol at the application layer.

**Section 2: Document the incident**

Several customers contacted the website's help desk indicating that, upon visiting it, they were asked to download and run a file containing access to new recipes. Their personal computers have run slowly ever since. The website owner tried to log in to the web server but noticed that they were locked out of their account.

The cybersecurity analyst used a sandbox environment to open the website without affecting the company's network. Then, the analyst ran tcpdump to capture network traffic packets generated by interacting with the website. The analyst was asked to download a file that claimed it would give access to free recipes, accepted the download, and executed it. The browser then redirected the analyst to a fake website (greatrecipesforme.com).

The cybersecurity analyst inspected the tcpdump log and observed that the browser initially requested the IP address of the yummyrecipesforme.com website. Once the connection to the website was established via the HTTP protocol, the analyst recalled having downloaded and executed the file. The logs showed a sudden change in network traffic when the browser requested a new IP address for the URL greatrecipesforme.com. The network traffic was then redirected to the new IP address of the greatrecipesforme.com website.

The senior cybersecurity professional analyzed the source code of the websites and the downloaded file. The analyst discovered that an attacker had manipulated the website to add code that incited users to download a malicious file disguised as a browser update. Since the website owner stated that they had been locked out of their administrator account, the team believes that the attacker used a brute-force attack to access the account and change the administrator password. The execution of the malicious file compromised the end users' computers.

**Section 3: Recommend remediation for brute-force attacks**

One security measure the team plans to implement to protect against brute-force attacks is to prevent the use of previous passwords. Given that the vulnerability that led to this attack was the attacker's ability to use a default password to log in, it is important that we prevent old passwords, such as default ones, from being used to reset the password. Another supporting measure is to require more frequent password updates, so that in case an unauthorized person learns the password, it is less likely that they can use it if it is updated soon. Finally, another useful solution is to implement two-factor authentication (2FA). 2FA authentication requires authentication via password and also confirming a one-time passcode (OTP) sent to your email or phone. Once the user confirms their identity through their login credentials and the OTP, they will have access to the system. Any malicious actor attempting a brute-force attack will likely not have access to the system because it requires additional authentication.

*C. Security Risk Assessment*

**Hardening tools and methods to implement**

Three hardening tools the organization can use to address the identified vulnerabilities include:
1.   Implementation of multi-factor authentication (MFA)
2.   Establishment and enforcement of strong password policies
3.   Regular firewall maintenance

MFA requires users to use more than one form of identification and credential verification before accessing an application. Some MFA methods include fingerprint scans, ID cards, PIN numbers, and passwords.

Password policies can be refined to include rules on password length, a list of acceptable characters, and a warning to discourage password sharing. They can also include rules on failed login attempts, such as the user losing network access after five failed attempts.

Firewall maintenance involves regularly reviewing and updating security configurations to anticipate potential threats.

**Recommendations**

Enforcing multi-factor authentication (**MFA**) adds an additional layer of security beyond just the password. This will reduce the likelihood of a malicious actor being able to access a network through a brute-force or related attack, as additional effort is required to authenticate in more than one way. MFA can also reduce the likelihood of people sharing passwords. Since the recipient of the shared password would need to possess additional authentication in addition to the password, MFA makes sharing passwords less useful, making them less likely to be shared.

Creating and enforcing a password policy within the company will make it increasingly difficult for malicious actors to access the network. Policies such as suspending an account after a certain number of login attempts can prevent successful brute-force attacks. Increasing password complexity, requiring more frequent updates, and not allowing passwords to be reused also helps curb the infiltration of malicious actors into the network.

Firewall maintenance should be done regularly. Network administrators must ensure that firewall rules reflect the most up-to-date standards for allowed and denied traffic. Traffic from suspicious sources should be included in a denied traffic list. Firewall rules should be updated whenever a security event occurs, especially an event that allows suspicious traffic into the network. This measure can be used to protect against various DoS (Denial-of-Service) and DDoS (Distributed Denial-of-Service) attacks.
