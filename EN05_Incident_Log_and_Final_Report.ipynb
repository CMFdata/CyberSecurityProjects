*B. Final report on unauthorized customer data access*

### *Incident Final Report*

**Executive Summary**

The organization experienced a security incident on April 19, 2026, at 19:20, during which an individual gained unauthorized access to personally identifiable customer information (PII) and financial information. Approximately 50,000 customer records were affected. The financial impact of the incident is estimated at $100,000 in direct costs and potential revenue losses. The incident is now closed, and a thorough investigation has been conducted.

**Chronology**

At approximately 3:13 PM on April 20, 2026, an employee received an email from an external address. The sender claimed to have stolen customer data. In exchange for not sharing the data on public forums, the sender demanded a ransom of $25,000 in cryptocurrency. The employee assumed the email was spam and deleted it.
On December 22, 2026, the same employee received another email from the same sender. This email included a sample of the stolen customer data and an increased ransom demand of $50,000.
That same day, the employee notified the security team, which launched an investigation into the incident. Between April 22 and 25, 2026, the security team focused on determining how the data was stolen and the extent of the breach.


**Timeline**

Approximately at 15:13 on April 20, 2026, an employee received an email from an external address. The sender claimed to have stolen customer data. In exchange for not releasing the data on public forums, the sender requested a cryptocurrency payment of USD 25,000. The employee assumed the email was spam and deleted it.
On December 22, 2026, the same employee received another email from the same sender. This email included a sample of the stolen customer data and an increased payment demand of USD 50,000.
On the same day, the employee notified the security team, who initiated the investigation into the incident. Between April 22 and April 25, 2026, the security team focused on determining how the data was stolen and the extent of the theft.

**Investigation**

The security team received the alert and traveled to the site to begin the investigation.
The root cause of the incident was identified as a vulnerability in the e-commerce web application. This vulnerability allowed the attacker to perform a forced browsing attack and access customer transaction data by modifying the order number included in the URL string of a purchase confirmation page. This vulnerability allowed the attacker to access customer purchase confirmation pages, exposing customer data, which was then collected and exfiltrated.
After confirming the web application vulnerability, the security team analyzed web application access logs. The logs indicated that the attacker accessed information from thousands of purchase confirmation pages.

**Response and Remediation**

The organization collaborated with the public relations department to disclose the data breach to its customers. Additionally, the organization offered free identity protection services to customers affected by the incident.
After the security team reviewed the associated web server logs, the cause of the attack became very clear. There was a single log source showing an exceptionally high volume of sequentially listed customer orders.

**Recommendations**

To prevent future recurrences, we are taking the following measures:

● Perform routine vulnerability scans and penetration tests.

● Implement the following access control mechanisms:

    ○ Implement a whitelist to allow access to a specific set of URLs and automatically block all requests outside this URL range.

    ○ Ensure that only authenticated users have authorized access to content.
